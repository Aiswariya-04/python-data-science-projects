## Introduction

Object-Oriented Programming (OOP) is a programming approach based on objects and classes.

OOP helps us organize programs in a structured and reusable manner.

In this task, three important OOP concepts are implemented:

- **Abstraction** – Hiding unnecessary implementation details.
- **Inheritance** – Reusing properties and methods from an existing class.
- **Polymorphism** – Allowing the same method to behave differently for different objects.

A banking example is used to understand these concepts practically.

```python
from abc import ABC, abstractmethod
```

## 1.Abstraction

Abstraction means hiding the internal implementation details and showing only the necessary functionality.

In Python, abstraction can be implemented using an abstract class.

Here, `BankAccount` is created as an abstract class.

The `withdraw()` method is declared as an abstract method. The child classes must provide their own implementation of this method.

```python
class BankAccount(ABC):
    
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    
    @abstractmethod
    def withdraw(self, amount):
        pass
    
    def display_balance(self):
        print("Account Holder:", self.account_holder)
        print("Current Balance: ₹", self.balance)
```

```python
try:
    account = BankAccount("Aiswariya", 10000)
except TypeError as e:
    print("Error:", e)
```

## 2. Inheritance 

Inheritance allows a child class to acquire the properties and methods of a parent class.

In this example:

- `BankAccount` is the parent class.
- `SavingsAccount` is a child class.
- `CurrentAccount` is another child class.

Both child classes inherit the properties and methods of `BankAccount`.

```python
class SavingsAccount(BankAccount):
    
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal successful from Savings Account.")
        else:
            print("Insufficient balance in Savings Account.")
```

```python
class CurrentAccount(BankAccount):
    
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal successful from Current Account.")
        else:
            print("Insufficient balance in Current Account.")
```

```python
savings = SavingsAccount("Aiswariya", 10000)

savings.display_balance()
```

## 3. Polymorphism

Polymorphism means "many forms".

In OOP, the same method can perform different actions depending on the object.

Here, both `SavingsAccount` and `CurrentAccount` have a method called `withdraw()`.

However, the same method can behave differently depending on the type of account.

This demonstrates polymorphism.

```python
savings = SavingsAccount("Aiswariya", 10000)
current = CurrentAccount("Ajay", 15000)

print("Savings Account:")
savings.withdraw(2000)
savings.display_balance()

print()

print("Current Account:")
current.withdraw(5000)
current.display_balance()
```

## Runtime polymorphism

Python allows the same method name to execute different implementations depending on the object.

In the following example, the variable `account` refers to different objects at different times.

The `withdraw()` method automatically calls the appropriate implementation.

```python
accounts = [
    SavingsAccount("Aiswariya", 10000),
    CurrentAccount("Ajay", 15000)
]

for account in accounts:
    print(type(account).__name__)
    account.withdraw(1000)
    account.display_balance()
    print()
```

```python
account.withdraw(1000)
```

```python
from abc import ABC, abstractmethod


# Abstract Parent Class
class BankAccount(ABC):
    
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    
    @abstractmethod
    def withdraw(self, amount):
        pass
    
    def display_balance(self):
        print("Account Holder:", self.account_holder)
        print("Current Balance: ₹", self.balance)


# Child Class 1
class SavingsAccount(BankAccount):
    
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal successful from Savings Account.")
        else:
            print("Insufficient balance in Savings Account.")


# Child Class 2
class CurrentAccount(BankAccount):
    
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal successful from Current Account.")
        else:
            print("Insufficient balance in Current Account.")


# Creating objects
savings_account = SavingsAccount("Aiswariya", 20000)
current_account = CurrentAccount("Bhavana", 30000)


# Savings Account
print("----- Savings Account -----")
savings_account.display_balance()
savings_account.withdraw(5000)
savings_account.display_balance()

print()

# Current Account
print("----- Current Account -----")
current_account.display_balance()
current_account.withdraw(10000)
current_account.display_balance()
```
