## 2.Create a list of numbers and print only even numbers using a for loop.

```python
numbers = (10,15,22,37,40,43,55,68)

print("Even Numbers:")

for num in numbers:
    if num % 2 == 0:
        print(num)
```

## 3. Write a program that handles division by zero using try-except.

```python
try:
    num = int(input("Enter Numerator:"))
    den = int(input("Enter Denominator"))
    result = num/den
    print("Result =",result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

## 4. Given a string, count how many vowels it has.

```python
text = input("Enter a string:")
count = 0
for letter in text.lower():
    if letter in "aeiou":
        count += 1
print("Number of vowels:",count)
```

## 5.Reverse a list without using built-in functions.

```python
numbers = [10,20,30,40,50]
reversed_list = []
for i in range(len(numbers)-1,-1,-1):
    reversed_list.append(numbers[i])
print("orginal List:",numbers)
print("Reversed List:",reversed_list)
```

## 9. Replace all missing values in a DataFrame with the column's mean

```python
data ={
    "Age": [20,22,24,None,None]
}
df = pd.DataFrame(data)
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```
