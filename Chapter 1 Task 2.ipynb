### 1.While loop example 1 ATM Balance Verification

```python
correct_pin = "1234"
entered_pin = ""

while entered_pin != correct_pin:
   entered_pin = input("Enter your ATM pin")

   if entered_pin != correct_pin:
    print("Incorrect PIN. Please try again.")

print("PIN verified successfully.Welcome!")
```

### Example 2 Employee Attendance Entry

```python
employee_count = 0
employee_name = ""

while employee_name.lower() != "stop":
    employee_name = input("Enter Employee Name")

    if employee_name.lower() != "stop":
        employee_count += 1
        print("Attendance recorded successfully.")

print("Attendance entry completed.")
print("Total Employees Present:",employee_count)
```

### UDF (With for loop & if condition)

Example 1 Eligible employees

```python
def eligible_employees():
    employee_ages = [19,24,23,34,32]

    for age in employee_ages:
        if age >= 18:
            print(age, "ELigible")

eligible_employees()
```

### Example 2: High Performing Employees

```python
def performance_scores():
    scores = [67,78,45,97,55]

    for score in scores:
        if score >= 75:
            print(score, "Excellent Performnace")

performance_scores()
```

### Break

```python
employees = ["Arshi","Indhu","Diya","Isha"]

for employee in employees:
      if employee == "Diya":
        print("Employee found")
        break

        print(employees)
```

### Example 2: Product Search

```python
products = ["Laptop","Mouse","keyboard","Printer"]

for product in products:
    if product == "Keyboard":
        print("Product Available")
        break

    print(product)
```

### continue

Example 1 Skip Absent Employee

```python
employees = ["Arun","Faisal","Harish","Janu"]

for employee in employees:
    if employee == "Absent":
        continue 

    print(employee)
```

### Example 2: Skip Failed Student

```python
marks = [85,40,92,30,76]

for mark in marks:
      if mark < 50:
          continue
print(mark)
```
